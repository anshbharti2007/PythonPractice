### `sum()` and `len()` Functions in Python

- **`len()` Function:**
  - It always starts with $0$.
  - If the dataset is empty, `len()` returns $0$.

- **Example:**
  - `set1 = {"apple", 5, "bat"}` assigns a variable.
  - `print(len(set1))` prints the number of items: $3$.

- **`sum()` Function:**
  - **Syntax:** `sum(iterable, start=0)`
  - Example:
    - `set1 = {5, 6, 10}` assigns a variable.
    - `print(sum(set1, 100))` adds the iterable and starts from base $100$. Result: $121$.
