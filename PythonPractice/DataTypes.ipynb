### Data Types in Python

- **Indexing:**
  - In Python, the index is the position of the item in a list. Index starts with $0$.
  - We can also use the index from the last item. It starts from the last item and goes from $-1$ to $-\infty$.

- **Examples:**
  - `vowel[0:3]` outputs $['a', 'e', 'i']$
  - `vowel[3:5]` outputs $['o', 'u']$

- **Books List Example:**
  - `Book = ["Harry Potter", "1984", "The Fault in Our Stars", "The Mom Test", "Life in Code"]`
  - Removing items:
    - `Book.remove("1984")`
    - `Book.pop(1)`
