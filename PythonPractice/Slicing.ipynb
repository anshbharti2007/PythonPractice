# **Slicing in Python and Numpy**

* sequence[start:stop:step]
- start : where to start slicing (inclusive)
- stop : where to stop slicing (exclusive) `important to note that the stop index is not included in the result`
- step : how many steps to jump while slicing (default is 1)


# **Slicing in Python Strings**

```python

text = "Hello World"
#       0123456789...

# Basic slicing
print(text[0:5])      # "Hello"    → index 0 to 4
print(text[6:11])     # "World"    → index 6 to 10
print(text[0:5:1])    # "Hello"    → step 1 (default)
print(text[0:5:2])    # "Hlo"      → every 2nd character

# Leaving parts empty
print(text[:5])       # "Hello"    → start from beginning
print(text[6:])       # "World"    → go till end
print(text[:])        # "Hello World" → full copy

# Step only
print(text[::2])      # "HloWrd"   → every 2nd char
print(text[::3])      # "HlWl"     → every 3rd char


```

# Negative Indexing in string

```python
text = "Hello World"
#             -5-4-3-2-1
#        -11 ........  -1

print(text[-5:])      # "World"    → last 5 characters
print(text[-5:-2])    # "Wor"      → index -5 to -3
print(text[-1])       # "d"        → last character
print(text[-3:-1])    # "rl"
print(text[:-5])      # "Hello "   → everything except last 5 characters
print(text[-6:-1])    # "Worl"     → from index -6 to -2

# Negative step = REVERSE
print(text[::-1])     # "dlroW olleH"  → reverse string
print(text[::-2])     # "drWolH"       → reverse, every 2nd

# In slicing, x is inclusive and y is exclusive, even with negative indices.
```

# Slicing in Python Lists

```python
nums = [10, 20, 30, 40, 50, 60, 70]
#        0   1   2   3   4   5   6

# Basic
print(nums[1:4])      # [20, 30, 40]
print(nums[:3])       # [10, 20, 30]
print(nums[4:])       # [50, 60, 70]
print(nums[:])        # [10, 20, 30, 40, 50, 60, 70] → copy

# With step
print(nums[::2])      # [10, 30, 50, 70] → every 2nd
print(nums[1::2])     # [20, 40, 60]     → start at 1, every 2nd
print(nums[1:6:2])    # [20, 40, 60]     → from 1 to 5, every 2nd

# Negative
print(nums[-3:])      # [50, 60, 70]
print(nums[-4:-1])    # [40, 50, 60]
print(nums[::-1])     # [70, 60, 50, 40, 30, 20, 10] → reversed
print(nums[::-2])     # [70, 50, 30, 10]

#Modifying 

nums = [10, 20, 30, 40, 50]

# Replace a section
nums[1:3] = [200, 300]
print(nums)           # [10, 200, 300, 40, 50]

# Delete a section
nums[1:3] = []
print(nums)           # [10, 40, 50]

# Insert without replacing
nums[1:1] = [99, 88]
print(nums)           # [10, 99, 88, 40, 50]
```

# Slicing in tuples 
* tuples are immutable, but we can still slice them to create new tuples.
* same as list slicing but can't modify the original tuple.

```python
# Same as list but CANNOT modify
t = (1, 2, 3, 4, 5)

print(t[1:4])         # (2, 3, 4)
print(t[::-1])        # (5, 4, 3, 2, 1)
print(t[::2])         # (1, 3, 5)

# t[1:3] = (9, 9)    ← ERROR! tuples are immutable will not work can be modified 
```
