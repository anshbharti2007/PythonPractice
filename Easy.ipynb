# 1: Contain Dublicates

```python
class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        s = set()
        for num in nums:
            if num in s:
                return True
            else:
                s.add(num)
        return False
```

# 2: Valid Anagram

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False
        counter = {}
        for char in s:
            counter[char] = counter.get(char, 0) + 1
        for char in t:
            counter[char] = counter.get(char, 0) - 1
        for  val in counter.values():
            if val != 0:
                return False 
        return True
```

# 3: Two Sum

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        store = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in store:
                return [store[complement], i] 
            store[num] = i 
        return []
```

# Valid Palindrome

```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        small = "".join(char.lower() for char in s if char.isalnum())
        return small == small[::-1]
```

# Best Time to Buy and Sell Stock

```python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        if not prices:
            return 0
            
        min_price = float('inf')      # or prices[0]
        max_profit = 0
        
        for price in prices:
            # Step 1: Update the minimum price seen so far
            if price < min_price:
                min_price = price
            
            # Step 2: Calculate potential profit if we sell today
            profit = price - min_price
            
            # Step 3: Update max_profit if today's profit is better
            if profit > max_profit:
                max_profit = profit
                
        return max_profit
```
