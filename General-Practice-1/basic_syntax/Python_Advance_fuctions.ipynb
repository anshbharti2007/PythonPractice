# Decorators in python
Decorators are a powerful and flexible way to modify the behavior of functions or classes in Python. They allow you to wrap another function or class, modifying its behavior without changing its source code. Decorators are often used for logging, access control, memoization, and more.

```python
# Decorator is a function that take a func and create  a new funtion which is called wrapper Then the wrapper uses the func in it with its funtions when called and then we return the wrapper funtion inside the decorator to decorator !

def decorator(func):
    def wrapper():
        print("i am about to execute a function..")
        func()
        print("i have executed a function !")
    return wrapper

# This is exactly equivalent to writing hello = decorator(hello)

@decorator 
def hello():
    print("Hello!")

# Now, whenever you call hello(), it usees the wrapper !
hello()
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(a):
            for i in range(n):
                func(a)
        return wrapper
    return decorator

@repeat(7)
def day_hello(a):
    print(f"Hello! {a}")

day_hello("Ansh")
```
