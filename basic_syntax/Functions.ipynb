# 🐍 Python Built-in Functions Guide

Welcome to the quick reference guide for Python's most commonly used built-in functions. To make learning easier, they are grouped by their primary purpose.

### 🔢 Math & Numbers
Functions used for mathematical operations and calculations.

| Function | Description | Example | Output |
| :--- | :--- | :--- | :--- |
| `abs()` | Returns the absolute (positive) value of a number. | `abs(-5)` | `5` |
| `max()` | Returns the largest item from an iterable or multiple arguments. | `max(1, 10, 5)` | `10` |
| `min()` | Returns the smallest item from an iterable or multiple arguments. | `min(1, 10, 5)` | `1` |
| `round()` | Rounds a number to a specified number of decimal places. | `round(3.14159, 2)` | `3.14` |
| `sum()` | Adds together all items of an iterable (like a list). | `sum([1, 2, 3])` | `6` |

### 🔄 Type Conversions
Functions used to convert data from one type to another.

| Function | Description | Example | Output |
| :--- | :--- | :--- | :--- |
| `bool()` | Converts a value to a boolean (`True` or `False`). | `bool(1)` | `True` |
| `dict()` | Creates a dictionary (key-value pairs). | `dict(a=1, b=2)` | `{'a': 1, 'b': 2}` |
| `float()` | Converts a value to a floating-point (decimal) number. | `float(5)` | `5.0` |
| `int()` | Converts a value to an integer (whole number). | `int(5.9)` | `5` |
| `list()` | Creates a list from an iterable. | `list("cat")` | `['c', 'a', 't']` |
| `set()` | Creates a set (a collection of unique items). | `set([1, 2, 2, 3])` | `{1, 2, 3}` |
| `str()` | Converts a value into a string (text). | `str(100)` | `'100'` |

### 📦 Iterables & Collections
Functions that help you work with sequences like lists, strings, and tuples.

| Function | Description | Example | Output |
| :--- | :--- | :--- | :--- |
| `all()` | Returns `True` if **all** items in an iterable are true. | `all([True, 1, "yes"])` | `True` |
| `any()` | Returns `True` if **any** item in an iterable is true. | `any([False, 0, "yes"])`| `True` |
| `enumerate()`| Adds a counter to an iterable, returning `(index, item)` pairs. | `list(enumerate(['a', 'b']))`| `[(0, 'a'), (1, 'b')]`|
| `filter()`| Keeps only the items that return `True` for a given function. | `list(filter(lambda x: x>0, [-1, 2]))`| `[2]` |
| `len()` | Returns the length (total number of items) of an object. | `len("hello")` | `5` |
| `map()` | Applies a specific function to every item in an iterable. | `list(map(len, ['hi', 'cat']))`| `[2, 3]` |
| `range()` | Generates a sequence of numbers (great for loops). | `list(range(3))` | `[0, 1, 2]` |
| `zip()` | Combines multiple iterables element-by-element. | `list(zip([1, 2], ['a', 'b']))`| `[(1, 'a'), (2, 'b')]`|

### 🛠️ Utilities & Helpers
Functions that help you debug, inspect, and display data.

| Function | Description | Example | Output |
| :--- | :--- | :--- | :--- |
| `dir()` | Returns a list of properties and methods associated with an object. | `dir([])` | `['append', 'clear', ...]` |
| `print()` | Prints output to the console for the user to see. | `print("Hello!")` | `Hello!` |
| `type()` | Tells you what data type an object is. | `type(3.14)` | `<class 'float'>` |

```python
class Drink:
    def __init__(self, name, brand, price):
        self.name = name
        self.brand = brand
        self.price = price
        self.next = None


d1 = Drink("Diet Coke", "Coca Cola", 60)
d2 = Drink("Lassi", "Amul", 20)
d3 = Drink("Black Coffee", "Davidoff", 500)

d1.next = d2
d2.next = d3

print("My Drink list")
current_drink = d1

while current_drink is not None:
    print(f"- {current_drink.brand} {current_drink.name} {current_drink.price}")
    current_drink = current_drink.next
```

```python
# ============================================================
#  PROJECT 1 — Set Operations Visualiser
#  IITM BS Math 1  |  Beginner Python
# ============================================================
#
#  HOW TO RUN:
#    python set_operations.py
#
#  WHAT YOU WILL PRACTISE:
#    - Python sets (they ARE math sets)
#    - Functions  (def)
#    - Loops      (for)
#    - Conditions (if / else)
#    - f-strings  (printing results neatly)
# ============================================================


# ------------------------------------------------------------
# STEP 1 — define your sets  (same as writing A = {1,3,5,7})
# ------------------------------------------------------------

A = {1, 3, 5, 7}
B = {2, 4, 6, 8}
U = {1, 2, 3, 4, 5, 6, 7, 8}   # Universal set


# ------------------------------------------------------------
# STEP 2 — a tiny helper so results print nicely
# ------------------------------------------------------------

def show(label, result):
    """Print a labelled set result."""
    print(f"  {label:<30} {sorted(result)}")
    #       ^ left-align label in 30 chars  ^ sort so order is predictable


# ------------------------------------------------------------
# STEP 3 — the six operations you know from class
# ------------------------------------------------------------

def set_operations(A, B, U):
    print("\n========== SET OPERATIONS ==========")
    print(f"  A = {sorted(A)}")
    print(f"  B = {sorted(B)}")
    print(f"  U = {sorted(U)}")
    print("-------------------------------------")

    show("A ∪ B  (union)",          A | B)
    show("A ∩ B  (intersection)",   A & B)
    show("A - B  (difference)",     A - B)
    show("B - A  (difference)",     B - A)
    show("A'     (complement of A)", U - A)
    show("B'     (complement of B)", U - B)

    print()
    print("  Verify De Morgan's 1st law:")
    left  = U - (A | B)          # (A ∪ B)'
    right = (U - A) & (U - B)    # A' ∩ B'
    print(f"    (A ∪ B)'  = {sorted(left)}")
    print(f"    A' ∩ B'   = {sorted(right)}")
    print(f"    Equal?    {left == right}")   # should always be True

    print()
    print("  Verify De Morgan's 2nd law:")
    left2  = U - (A & B)         # (A ∩ B)'
    right2 = (U - A) | (U - B)   # A' ∪ B'
    print(f"    (A ∩ B)'  = {sorted(left2)}")
    print(f"    A' ∪ B'   = {sorted(right2)}")
    print(f"    Equal?    {left2 == right2}")


# ------------------------------------------------------------
# STEP 4 — Cartesian product  (A × B)
#           This is where relation R lives!
# ------------------------------------------------------------

def cartesian_product(A, B):
    print("\n========== CARTESIAN PRODUCT A×B ==========")
    product = set()
    for a in A:
        for b in B:            # nested loop = every (a,b) pair
            product.add((a, b))

    print(f"  Total pairs: {len(product)}   (should be {len(A)} × {len(B)} = {len(A)*len(B)})")
    for pair in sorted(product):
        print(f"    {pair}")
    return product


# ------------------------------------------------------------
# STEP 5 — Build relation R  where  a < b
#           (the exact relation from your problem!)
# ------------------------------------------------------------

def build_relation(A, B):
    print("\n========== RELATION  R : a < b ==========")

    R = set()
    for a in A:
        for b in B:
            if a < b:          # the condition that defines R
                R.add((a, b))

    print(f"  Pairs in R:  {sorted(R)}")
    print(f"  |R| = {len(R)}   (cardinality)")

    # Domain  = all first elements
    domain = {a for a, b in R}     # set comprehension — new concept!
    print(f"  Domain of R:       {sorted(domain)}")

    # Range (image) = all second elements
    range_ = {b for a, b in R}
    print(f"  Range of R:        {sorted(range_)}")

    return R


# ------------------------------------------------------------
# STEP 6 — Inverse relation  R⁻¹
#           just flip every pair (a,b) → (b,a)
# ------------------------------------------------------------

def inverse_relation(R):
    print("\n========== INVERSE RELATION  R⁻¹ ==========")

    R_inv = set()
    for a, b in R:
        R_inv.add((b, a))      # swap the two elements

    print(f"  Pairs in R⁻¹: {sorted(R_inv)}")

    domain_inv = {x for x, y in R_inv}
    range_inv  = {y for x, y in R_inv}

    print(f"  Domain of R⁻¹:  {sorted(domain_inv)}")
    print(f"  Range  of R⁻¹:  {sorted(range_inv)}")

    # Verify option (a) from your problem
    domain_R = {a for a, b in R}
    print()
    print("  Verify: Domain(R) == Range(R⁻¹) ?")
    print(f"    Domain(R)    = {sorted(domain_R)}")
    print(f"    Range(R⁻¹)  = {sorted(range_inv)}")
    print(f"    Equal?       {domain_R == range_inv}")   # True ✓


# ------------------------------------------------------------
# STEP 7 — Power set  P(A)
#           all possible subsets of A
# ------------------------------------------------------------

def power_set(A):
    print("\n========== POWER SET  P(A) ==========")

    A_list = list(A)           # sets can't be indexed, lists can
    n      = len(A_list)
    result = []

    # Trick: a set with n elements → 2^n subsets
    # Each number from 0 to 2^n - 1 in binary tells us which elements to pick
    for i in range(2 ** n):
        subset = set()
        for j in range(n):
            if i & (1 << j):   # bitwise: is the j-th bit of i set?
                subset.add(A_list[j])
        result.append(subset)

    print(f"  |A| = {n}  →  |P(A)| = 2^{n} = {2**n}")
    print(f"  All subsets:")
    for s in result:
        print(f"    {sorted(s)}")

    return result


# ------------------------------------------------------------
# STEP 8 — run everything
# ------------------------------------------------------------

if __name__ == "__main__":
    set_operations(A, B, U)
    cartesian_product(A, B)
    R = build_relation(A, B)
    inverse_relation(R)
    power_set(A)

    print("\n========== DONE ==========")
    print("  Try changing A, B, U at the top of the file")
    print("  and re-run. Does R change? Does |P(A)|?")
```
