```python
# problem 1
fruits = ["apple",  "banana", "cherry"] #creating a list
fruits.append("Goldenfish") # add a item at last of list 
fruits.insert(5, "Time") # adding a item on an index 
fruits.remove("Goldenfish") # removing specific value 
fruits.pop() # removing the last item in the list
fruits.reverse() # will reverse the list 
fruits.sort() # sort the list based on ascending order 
print(fruits) # print the list
```

```python
# set and set methods 
balls = {"pink", "Red", "White"} # set example 
cat = {"Larry", "John", "White"}
balls.add("Black") # add black at random 
balls.remove("pink") # remove pink if not found error 
balls.discard(9) # remove 9 if not there error 
balls.pop() # remove a random element
print(balls) # print function
print(balls.intersection(cat))
print(balls.union(cat))
print(balls.difference(cat))
print(balls.clear())
```
