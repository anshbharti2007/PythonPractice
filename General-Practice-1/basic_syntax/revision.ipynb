# find largest number without max in a list 


```python
nums = [3, 1, 4, 1, 5, 9]
if not nums:
    print("the list is empty")
else:
    current = nums[0]
for num in nums:
    if num > current:
        current = num 
print(f"The largest number in nums is {current}")
    
```

# Find the second largest number in a list 

```python

nums = [3, 1, 4, 1, 5, 9]
max = nums[0]
secondmax = 0
for num in nums:
    if num > max:
        secondmax = num
        max = num
for num in nums:
    if num > secondmax and secondmax != max:
        secondmax = num 
print(f"The second largest number is {secondmax}")
#the largest sum
```
